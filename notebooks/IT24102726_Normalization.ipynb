```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("C:\\Users\\Ashen\\Documents\\GitHub\\Employee-Attrition-Prediction-ML-Model\\data\\Employee Attrition Prediction Dataset.csv")

print("Shape:", df.shape)
df.head()
```

```python
# Select numeric columns
numeric_cols = df.select_dtypes(include=['int64','float64']).columns.tolist()
print("Numeric columns:", numeric_cols)
```

```python
scaler = StandardScaler()
df_scaled = df.copy()

# Apply to selected columns
scale_cols = ["Years at Company", "Monthly Income", "Number of Promotions", "Distance from Home", "Number of Dependents", "Company Tenure"]
df_scaled[scale_cols] = scaler.fit_transform(df_scaled[scale_cols])

df_scaled.head()
```

```python
# Example: Years at Company distribution before and after scaling
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
sns.histplot(df["Years at Company"], bins=30, kde=True, color="blue")
plt.title("Before Scaling")

plt.subplot(1,2,2)
sns.histplot(df_scaled["Years at Company"], bins=30, kde=True, color="green")
plt.title("After Standard Scaling")

plt.tight_layout()
plt.show()
```
