```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("C:\\Users\\Ashen\\Desktop\\aiml\\Employee Attrition Prediction Dataset.csv")

print("Shape:", df.shape)
df.head()
```

```python
# Select numeric columns
numeric_cols = df.select_dtypes(include=['int64','float64']).columns.tolist()
print("Numeric columns:", numeric_cols)
```

```python
scaler = StandardScaler()

# Apply to selected columns
scale_cols = ["Age","Years at Company", "Monthly Income", "Number of Promotions", "Distance from Home", "Number of Dependents", "Company Tenure"]
df[scale_cols] = scaler.fit_transform(df[scale_cols])

df.head()
```

```python
#Age after scaling
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
sns.histplot(df["Age"], bins=30, kde=True, color="blue")
plt.title("After Scaling")

plt.tight_layout()
plt.show()
```

```python
#Years at Company distribution after scaling
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
sns.histplot(df["Years at Company"], bins=30, kde=True, color="blue")
plt.title("After Scaling")

plt.tight_layout()
plt.show()
```

```python
# Monthly Income distribution after scaling
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
sns.histplot(df["Monthly Income"], bins=30, kde=True, color="blue")
plt.title("After Scaling")

plt.tight_layout()
plt.show()
```

```python
# Years at Company distribution after scaling
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
sns.histplot(df["Number of Promotions"], bins=30, kde=True, color="blue")
plt.title("after Scaling")


plt.tight_layout()
plt.show()
```

```python
# Distance from Home distribution after scaling
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
sns.histplot(df["Distance from Home"], bins=30, kde=True, color="blue")
plt.title("After Scaling")

plt.tight_layout()
plt.show()
```

```python
# Number of Dependents distribution after scaling
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
sns.histplot(df["Number of Dependents"], bins=30, kde=True, color="blue")
plt.title("After Scaling")

plt.tight_layout()
plt.show()
```

```python
# Company Tenure distribution after scaling
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
sns.histplot(df["Company Tenure"], bins=30, kde=True, color="blue")
plt.title("After Scaling")

plt.tight_layout()
plt.show()
```
